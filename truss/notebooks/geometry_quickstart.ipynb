# Trellis Quickstart: 3D Geometry and Sessions

This notebook demonstrates using the **Trellis** Python API to initialize an interactive session, parse 3D polygon meshes (Wavefront OBJ), and inspect point clouds (PTS).

## 1. Import Trellis and Check Version

Ensure `trellis` is installed in your Jupyter kernel's environment (`pip install ./python` or `maturin develop`).

```python
import trellis
print(f"Trellis version: {trellis.__version__}")
```

## 2. Initialize a Session

A `Session` provides the primary gateway to Trellis's native services and shared runtime state.

```python
session = trellis.Session()
print(session)
```

## 3. Parse and Inspect a Polygonal Mesh (Wavefront OBJ)

We can parse OBJ data directly from a string or load it from a file path using `session.geometry.load()`.

```python
obj_content = """v -0.5 -0.5 0.0
v 0.5 -0.5 0.0
v 0.0 0.5 0.0
f 1 2 3
"""

mesh = session.parse_obj(obj_content)
print(f"Asset repr:    {mesh}")
print(f"Vertex count:  {mesh.vertex_count}")
print(f"Face count:    {mesh.face_count}")
print(f"Is point cloud:{mesh.is_point_cloud}")
print(f"Positions:     {mesh.vertex_positions()}")
print(f"Face indices:  {mesh.faces()}")
```

## 4. Parse Point Cloud Data (PTS)

Trellis natively supports `.pts` point cloud data with positions, intensities, and per-vertex colors.

```python
pts_content = """3
0.0 0.0 0.0 255 255 0 0
1.0 0.0 0.0 200 0 255 0
0.0 1.0 0.0 150 0 0 255
"""

cloud = session.parse_pts(pts_content)
print(f"Point count:   {cloud.point_count}")
print(f"Is point cloud:{cloud.is_point_cloud}")
print(f"Colors:        {cloud.vertex_colors()}")
```

## 5. Error Handling

Trellis converts native Rust errors into standard Python exceptions (`FileNotFoundError`, `ValueError`).

```python
try:
    session.load_geometry("missing_model.obj")
except FileNotFoundError as err:
    print(f"Caught expected exception: {err}")

try:
    session.parse_obj("invalid obj content")
except ValueError as err:
    print(f"Caught expected syntax error: {err}")
```
